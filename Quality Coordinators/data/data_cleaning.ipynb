```python
import pandas as pd
import numpy as np
import os
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("adityaramachandran27/world-air-quality-index-by-city-and-coordinates")
print("Path to dataset files:", path)
print(os.listdir(path))
csv_file = "AQI and Lat Long of Countries.csv"  
file_path = os.path.join(path, csv_file)
df = pd.read_csv(file_path)
```

```python
# check for null values
df.isnull().sum()
```

```python
# Drop rows with any null values
df.dropna(inplace=True)
```

```python
# Get a sum of rows with duplicate city names
df['City'].duplicated().sum()
```

```python
# Drop rows with duplicated city names but keep the first duplicate
df.drop_duplicates(subset=['City'], keep='first', inplace=True) 
df
```

```python
# Reset index
df.reset_index(drop=True)
df
```

```python
df.columns
```

```python
# Change column names
df.columns = ['Country','City','AQI_value','AQI_category','CO_AQI_value','CO_AQI_category',
              'Ozone_AQI_value','Ozone_AQI_category','NO2_AQI_value','NO2_AQI_category',
              'PM2_5_AQI_value','PM2_5_AQI_category','latitude','longitude']
df.head()
```

```python
# Export cleansed data
df.to_csv('world_AQI_cleaned.csv', index=False)
```
