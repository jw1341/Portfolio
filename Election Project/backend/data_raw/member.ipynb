```python
import random
import string
```

```python
# Function to generate a secure password (excluding commas)
def generate_password(length=15):
    if length < 10:  # Ensure minimum length for required characters
        raise ValueError("Password length must be at least 10")

    # Define characters excluding commas
    special_chars = string.punctuation.replace(",", "")  # Remove commas
    all_chars = string.ascii_letters + string.digits + special_chars

    # Ensure at least one uppercase, one number, and one special character (excluding comma)
    upper = random.choice(string.ascii_uppercase)
    lower = random.choice(string.ascii_lowercase)
    digit = random.choice(string.digits)
    special = random.choice(special_chars)

    # Generate remaining random characters
    remaining_length = length - 3
    other_chars = ''.join(random.choices(all_chars, k=remaining_length))

    # Shuffle the password to ensure randomness
    password = list(upper + lower + digit + special + other_chars)
    random.shuffle(password)

    return ''.join(password)

# Add 'password' column with generated passwords
df['password'] = [generate_password() for _ in range(len(df))]

# Display the updated DataFrame
df
```
